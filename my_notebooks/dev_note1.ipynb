```python
import sys
sys.path.append('../python')

import urllib
import json
```

```python
spot_url = "https://api.binance.com/api/v3/exchangeInfo"
cm_url = "https://dapi.binance.com/dapi/v1/exchangeInfo"
um_url = "https://fapi.binance.com/fapi/v1/exchangeInfo"
```

```python
response = urllib.request.urlopen(spot_url).read()
response = json.loads(response)
```

```python
response.keys()
```

```python
response['symbols'][0]
```

```python
def symbol2pair(response):
    ret = {}
    for item in response['symbols']:
        symbol = item['symbol']
        base = item['baseAsset']
        quote = item['quoteAsset']
        ret[symbol] = (base, quote)
    return ret
```

```python
symbol2pair_dict = symbol2pair(response)
```

```python
for key, (base, quote) in symbol2pair_dict.items():
    if base == 'USD' or quote == 'USD':
        print(key, base, quote)
```
